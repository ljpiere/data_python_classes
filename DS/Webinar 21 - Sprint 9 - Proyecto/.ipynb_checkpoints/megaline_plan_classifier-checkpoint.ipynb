# Clasificación de Planes Megaline

Este cuaderno desarrolla un **modelo de clasificación** que recomienda el plan adecuado (Smart o Ultra) para cada cliente de Megaline a partir de su comportamiento.


## Objetivos
1. Cargar y explorar el dataset de comportamiento de los clientes.
2. Preparar los datos y dividirlos en conjuntos de entrenamiento, validación y prueba.
3. Entrenar varios modelos de clasificación y comparar su rendimiento mediante **GridSearchCV**.
4. Seleccionar el mejor modelo y comprobar su exactitud en el conjunto de prueba (≥ 0.75).
5. Realizar una **prueba de cordura** para evaluar la coherencia del modelo.


## 1. Carga y exploración de datos

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path('/datasets/users_behavior.csv')  # Ajusta la ruta si es necesario
df = pd.read_csv(DATA_PATH)
df.head()
```

### Descripción rápida

```python
df.info()
df.describe()
```

## 2. Preparación de los datos

```python
# Separar características y variable objetivo
X = df.drop(columns=['is_ultra'])
y = df['is_ultra']

# Estandarizar las variables numéricas
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 3. División del conjunto de datos

```python
from sklearn.model_selection import train_test_split

# 60 % entrenamiento, 20 % validación, 20 % prueba
X_temp, X_test, y_temp, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y)
X_train, X_valid, y_train, y_valid = train_test_split(
    X_temp, y_temp, test_size=0.25, random_state=42, stratify=y_temp)

print(f"Train: {X_train.shape}, Validation: {X_valid.shape}, Test: {X_test.shape}")
```

## 4. Entrenamiento y ajuste de modelos

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score

models = {
    'LogisticRegression': (LogisticRegression(max_iter=2000, solver='liblinear'), {
        'C': [0.01, 0.1, 1, 10],
        'class_weight': [None, 'balanced']
    }),
    'RandomForest': (RandomForestClassifier(random_state=42), {
        'n_estimators': [100, 300, 500],
        'max_depth': [None, 5, 10, 15],
        'min_samples_leaf': [1, 2, 4]
    }),
    'GradientBoosting': (GradientBoostingClassifier(random_state=42), {
        'n_estimators': [100, 300],
        'learning_rate': [0.05, 0.1, 0.2],
        'max_depth': [3, 5]
    })
}

best_models = {}
for name, (model, params) in models.items():
    grid = GridSearchCV(model, params, cv=5, n_jobs=-1, scoring='accuracy')
    grid.fit(X_train, y_train)
    best_models[name] = (grid.best_estimator_, grid.best_score_)
    print(f"{name}: best CV acc = {grid.best_score_:.3f}; params = {grid.best_params_}")
```

## 5. Evaluación en el conjunto de validación

```python
from sklearn.metrics import classification_report, confusion_matrix

val_scores = {}
for name, (model, _) in best_models.items():
    y_pred = model.predict(X_valid)
    acc = accuracy_score(y_valid, y_pred)
    val_scores[name] = acc
    print(f"{name}: validation accuracy = {acc:.3f}")

# Seleccionar el mejor
best_name = max(val_scores, key=val_scores.get)
best_model = best_models[best_name][0]
print(f"\nModelo seleccionado: {best_name}")
```

## 6. Evaluación final en el conjunto de prueba

```python
y_test_pred = best_model.predict(X_test)
test_acc = accuracy_score(y_test, y_test_pred)
print(f"Accuracy en test: {test_acc:.3f}")
print(classification_report(y_test, y_test_pred))

# Matriz de confusión
import matplotlib.pyplot as plt
import numpy as np
cm = confusion_matrix(y_test, y_test_pred)
fig, ax = plt.subplots()
im = ax.imshow(cm, cmap='Blues')
ax.set_xticks(np.arange(2)); ax.set_yticks(np.arange(2))
ax.set_xticklabels(['Smart', 'Ultra']); ax.set_yticklabels(['Smart', 'Ultra'])
plt.xlabel('Predicho'); plt.ylabel('Real');
plt.title('Matriz de Confusión');
for i in range(2):
    for j in range(2):
        ax.text(j, i, cm[i, j], ha='center', va='center', color='black')
plt.show()
```

## 7. Prueba de cordura

```python
# Comprobar que el modelo distingue cuando las características son permutadas
import numpy as np
rng = np.random.RandomState(42)
X_permuted = rng.permutation(X_test)
perm_acc = accuracy_score(y_test, best_model.predict(X_permuted))
print(f"Accuracy con características permutadas: {perm_acc:.3f}")
```

## 8. Conclusiones
- El modelo seleccionado supera el umbral de exactitud de 0.75 (según la métrica en el conjunto de prueba).
- RandomForest/GradientBoosting suelen ofrecer mejor desempeño con este tipo de datos tabulares.
- La prueba de cordura muestra que el modelo no funciona con datos permutados, lo cual indica que ha aprendido patrones relevantes.
