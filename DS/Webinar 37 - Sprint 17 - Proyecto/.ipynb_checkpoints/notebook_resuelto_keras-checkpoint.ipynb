
# Predicción de Edad a partir de Rostros — Notebook (Keras)

Este notebook implementa **de principio a fin** un flujo de trabajo para predecir edad con **Keras (TensorFlow)** usando un dataset de imágenes y un archivo `labels.csv` con las columnas:

- `file_name`: nombre del archivo de imagen (ej. `000123.jpg`)
- `real_age`: edad real (entero)

## Estructura esperada
- Carpeta con imágenes: `/final_files`
- CSV de etiquetas: `labels.csv` (en el mismo directorio que este notebook)

> **Nota**: Este notebook **no** se ejecuta aquí en el entorno actual por ausencia de TensorFlow; está listo para correr **en tu PC local**.


## Inicialización

```python

import os, json, math, random, gc
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Si usas GPU y TensorFlow local, descomenta:

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models, optimizers, callbacks

BASE_DIR = Path(__file__).parent if "__file__" in globals() else Path.cwd()

print("TensorFlow:", tf.__version__)
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

# Paths principales
IMAGES_DIR = (BASE_DIR / "final_files").resolve()   # <-- AJUSTA si tus imágenes están en otra ruta
LABELS_CSV = Path("labels.csv")     # <-- AJUSTA si tu CSV está en otra ruta

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 20

assert LABELS_CSV.exists(), f"No se encontró {LABELS_CSV.resolve()}"
print("Usando imágenes en:", IMAGES_DIR.resolve())
```

## Carga los datos

El conjunto de datos se almacena en la carpeta `/final_files/` 

```python
labels = pd.read_csv(LABELS_CSV)
assert {'file_name','real_age'}.issubset(labels.columns), "El CSV debe tener columnas file_name y real_age"

# Construye ruta absoluta a cada imagen
labels['filepath'] = labels['file_name'].apply(lambda x: str(IMAGES_DIR / x))
print(labels.head())
print("Total filas:", len(labels))
```

```python
# Filtrar filas cuya imagen no exista (por seguridad)
exists_mask = labels['filepath'].apply(lambda p: Path(p).exists())
if exists_mask.sum() < len(labels):
    print("Aviso: Hay filas cuyas imágenes no existen. Se filtrarán.")
labels = labels.loc[exists_mask].reset_index(drop=True)
print("Total tras filtrar:", len(labels))
```

## Modelado

Define las funciones necesarias para entrenar todo el pipeline; crea un solo script que las contenga todas junto con la sección de entrenamiento y evaluación.

```python

import pandas as pd
from sklearn.model_selection import train_test_split

def load_train(path: str, img_size=(224,224), batch_size=32, validation_split=0.2, seed=42):
    """
    Carga la parte de entrenamiento usando flow_from_dataframe con augmentación.
    
    Args:
        path: ruta a las imágenes (carpeta)
        img_size: tamaño target (alto, ancho)
        batch_size: tamaño de batch
        validation_split: proporción para validación
        seed: semilla reproducible
    
    Returns:
        train_gen_flow: generator de entrenamiento
        val_gen_flow: generator de validación
        train_df, val_df: dataframes usados
    """
    assert isinstance(path, (str, Path))
    images_dir = Path(path)
    df = labels.copy()
    # split train/val
    train_df, val_df = train_test_split(df, test_size=validation_split, random_state=seed)

    train_datagen = ImageDataGenerator(
        rescale=1./255.0,
        rotation_range=12,
        width_shift_range=0.08,
        height_shift_range=0.08,
        zoom_range=0.12,
        horizontal_flip=True,
    )
    val_datagen = ImageDataGenerator(rescale=1./255.0)

    train_gen_flow = train_datagen.flow_from_dataframe(
        dataframe=train_df,
        x_col='filepath',
        y_col='real_age',
        target_size=img_size,
        batch_size=batch_size,
        class_mode='raw',
        shuffle=True,
        seed=seed,
    )
    val_gen_flow = val_datagen.flow_from_dataframe(
        dataframe=val_df,
        x_col='filepath',
        y_col='real_age',
        target_size=img_size,
        batch_size=batch_size,
        class_mode='raw',
        shuffle=False
    )
    return train_gen_flow, val_gen_flow, train_df, val_df
```

```python

def load_test(path: str, img_size=(224,224), batch_size=32, df: pd.DataFrame=None):
    """
    Carga la parte de test. 
    - Si se pasa un df, lo usa (con columnas filepath y real_age opcionalmente).
    - Si no, por defecto usa el resto del dataset.
    
    Retorna:
        test_gen_flow, test_df
    """
    if df is None:
        test_df = labels.copy()
    else:
        test_df = df.copy()

    test_datagen = ImageDataGenerator(rescale=1./255.0)
    test_gen_flow = test_datagen.flow_from_dataframe(
        dataframe=test_df,
        x_col='filepath',
        y_col='real_age' if 'real_age' in test_df.columns else None,
        target_size=img_size,
        batch_size=batch_size,
        class_mode='raw' if 'real_age' in test_df.columns else None,
        shuffle=False
    )
    return test_gen_flow, test_df
```

```python

def create_model(input_shape):
    """
    Define una CNN sencilla para regresión de edad.
    Evitamos pesos pre-entrenados para no requerir descargas.
    """
    inputs = layers.Input(shape=input_shape)
    x = layers.Conv2D(16, 3, activation='relu', padding='same')(inputs)
    x = layers.MaxPooling2D()(x)
    x = layers.Conv2D(32, 3, activation='relu', padding='same')(x)
    x = layers.MaxPooling2D()(x)
    x = layers.Conv2D(64, 3, activation='relu', padding='same')(x)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Dense(64, activation='relu')(x)
    outputs = layers.Dense(1, activation='linear')(x)
    model = models.Model(inputs, outputs)
    model.compile(optimizer=optimizers.Adam(1e-3), loss='mae', metrics=['mae'])
    return model
```

```python

def plot_history(history):
    hist = history.history
    plt.figure()
    plt.plot(hist['loss'], label='loss')
    if 'val_loss' in hist: plt.plot(hist['val_loss'], label='val_loss')
    plt.xlabel('Epoch')
    plt.ylabel('MAE')
    plt.title('Entrenamiento (MAE)')
    plt.legend()
    plt.show()
```

### Entrenamiento

```python

train_gen, val_gen, train_df, val_df = load_train(IMAGES_DIR, img_size=IMG_SIZE, batch_size=BATCH_SIZE, validation_split=0.2, seed=SEED)
model = create_model(input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3))
model.summary()

cb = [
    callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
    callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, verbose=1),
    callbacks.ModelCheckpoint('best_model.keras', save_best_only=True, monitor='val_loss', verbose=1)
]

history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=EPOCHS,
    callbacks=cb,
    verbose=1
)

plot_history(history)
```

### Evaluación

```python
# Evaluación en validación
val_metrics = model.evaluate(val_gen, return_dict=True)
print("Métricas validación:", val_metrics)

# Predicciones y análisis de errores
preds = model.predict(val_gen).ravel()   # equivalente a reshape(-1)
true_y = val_df['real_age'].values

from sklearn.metrics import mean_absolute_error, mean_squared_error
mae = mean_absolute_error(true_y, preds)
rmse = mean_squared_error(true_y, preds, squared=False)
print(f"MAE: {mae:.3f} | RMSE: {rmse:.3f}")

plt.figure()
plt.scatter(true_y, preds, alpha=0.5)
plt.xlabel("Edad real")
plt.ylabel("Edad predicha")
plt.title("Validación - Real vs Predicho")
plt.show()

# Guarda artefactos
import json
# 1) Formato nativo Keras (recomendado Keras 3)
model.save("final_model.keras")

# 2) SavedModel para TF Serving / TFLite (carpeta)
model.export("final_model_saved")

# 3) CSV con predicciones
out_df = pd.DataFrame({
    'file_name': val_df['file_name'].values,
    'true_age': true_y,
    'pred_age': preds
})
out_df.to_csv('val_predictions.csv', index=False)

print("Artefactos guardados: final_model.keras, final_model_saved/, val_predictions.csv")
```

### Inferencia en nuevas imágenes

```python

def predict_single_image(path, img_size=IMG_SIZE):
    from tensorflow.keras.preprocessing import image
    img = image.load_img(path, target_size=img_size)
    x = image.img_to_array(img)/255.0
    x = np.expand_dims(x, axis=0)
    pred = model.predict(x)[0][0]
    return pred

# Ejemplo (ajusta el nombre de archivo existente en /final_files)
# print(predict_single_image(IMAGES_DIR / '000123.jpg'))
```
