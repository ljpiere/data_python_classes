## 1) Carga del dataset e inspección

**Qué haremos y por qué:**  
- Leer `taxi.csv`.
- Detectar y **parsear** la columna de fechas (`datetime` detectada heurísticamente).
- Ordenar por tiempo y establecer índice temporal.
- Confirmar periodicidad horaria; si no, **re-muestrear** a 1H y completar huecos.


```
# --- Setup e imports ---
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# --- Carga de datos ---
data_path = Path("/mnt/data/taxi.csv")
df = pd.read_csv(data_path)

# Heurísticas de nombres; puedes ajustarlos manualmente si difieren
date_col = "datetime"
target_col = "num_orders"

# Intento de parseo de fechas
try:
    df[date_col] = pd.to_datetime(df[date_col], infer_datetime_format=True, utc=False, errors="raise")
except Exception as e:
    # Reintento laxo; si falla, indica el nombre correcto de columna de fecha
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
    assert df[date_col].notna().any(), "No se pudo parsear la columna temporal; revisa 'date_col'."

# Orden y set index
df = df.sort_values(date_col).reset_index(drop=True)
df = df.set_index(date_col)

print("Rango temporal:", df.index.min(), "→", df.index.max())
print("Frecuencia inferida:", pd.infer_freq(df.index))

# Si no está en frecuencia horaria, re-muestreamos a 1H
df_res = df.copy()
if pd.infer_freq(df_res.index) != "H":
    df_res = df_res.resample("H").sum(numeric_only=True)

# Completar el calendario horario para evitar huecos en el índice
full_idx = pd.date_range(df_res.index.min(), df_res.index.max(), freq="H")
df_res = df_res.reindex(full_idx)

# Selección de objetivo: si no existe, levanta error
assert target_col in df_res.columns, f"No se encuentra la columna objetivo '{target_col}'. Columnas: {list(df_res.columns)}"

# Tratamiento simple de faltantes en target: 0 -> si representa conteo; de lo contrario forward fill
if df_res[target_col].isna().mean() > 0:
    # Heurística: si parece conteo no negativo, llenamos con 0; si no, ffill
    if (df_res[target_col] >= 0).all():
        df_res[target_col] = df_res[target_col].fillna(0)
    else:
        df_res[target_col] = df_res[target_col].fillna(method="ffill").fillna(method="bfill")

df_res.head(3)
```

### EDA mínimo y chequeos

- `info()` para verificar tipos y nulos.  
- Duplicados temporales.  
- Valores negativos/atípicos en la variable objetivo.  
- Visualización de la serie y medias móviles (no fijar estilos/colores).


```
# --- EDA ---
print(df_res.info())
dups = df_res.index.duplicated().sum()
print("Duplicados de índice temporal:", dups)

desc = df_res.describe()
print(desc)

# Chequeo rápido de negativos
neg_ratio = (df_res[target_col] < 0).mean()
print(f"Proporción de valores negativos en '{target_col}': {neg_ratio:.4f}")

# Visualización básica
plt.figure(figsize=(12,4))
plt.plot(df_res.index, df_res[target_col])
plt.title("Serie temporal - objetivo")
plt.xlabel("Tiempo")
plt.ylabel(target_col)
plt.show()

# Medias móviles
for w in [3, 24, 168]:
    plt.figure(figsize=(12,3))
    plt.plot(df_res.index, df_res[target_col].rolling(w, min_periods=1).mean())
    plt.title(f"Media móvil {w}h")
    plt.xlabel("Tiempo")
    plt.ylabel(f"{target_col}_ma{w}")
    plt.show()
```

## 2) Ingeniería de características

**Qué haremos y por qué:**  
- **Lags** (1, 2, 3, 24): capturan dependencia temporal.  
- **Rolling stats**: medias móviles y desviación (3h, 24h, 168h).  
- **Calendario**: hora del día, día de la semana, fin de semana.  
- Evitamos incluir el valor actual del target para no generar **fuga**.


```
# --- Funciones para features ---
def make_features(df_ts, target, lags=(1,2,3,24), roll_windows=(3,24,168)):
    df_feat = pd.DataFrame(index=df_ts.index)
    df_feat[target] = df_ts[target]

    # Lags
    for l in lags:
        df_feat[f"{target}_lag{l}"] = df_ts[target].shift(l)

    # Rolling stats
    for w in roll_windows:
        df_feat[f"{target}_ma{w}"] = df_ts[target].shift(1).rolling(w).mean()
        df_feat[f"{target}_std{w}"] = df_ts[target].shift(1).rolling(w).std()

    # Calendario
    df_feat["hour"] = df_ts.index.hour
    df_feat["dow"] = df_ts.index.dayofweek
    df_feat["is_weekend"] = (df_feat["dow"] >= 5).astype(int)

    return df_feat

df_feat = make_features(df_res, target_col)
df_feat = df_feat.dropna()  # perderemos las primeras filas por lags/rolling
print("Shape con features:", df_feat.shape)
df_feat.head(3)
```

## 3) Split temporal (Train/Valid/Test) + Baselines

- Usamos cortes por tiempo: 80% train, 10% valid, 10% test (editables).  
- Baselines: **naive** (valor previo) y **media móvil** (24h).  
- Métrica: **RMSE**.


```
# --- Split temporal ---
def time_split(df_, train_ratio=0.8, valid_ratio=0.1):
    n = len(df_)
    n_train = int(n * train_ratio)
    n_valid = int(n * valid_ratio)
    train = df_.iloc[:n_train]
    valid = df_.iloc[n_train:n_train+n_valid]
    test  = df_.iloc[n_train+n_valid:]
    return train, valid, test

train, valid, test = time_split(df_feat, 0.8, 0.1)

features = [c for c in df_feat.columns if c != target_col]
X_train, y_train = train[features], train[target_col]
X_valid, y_valid = valid[features], valid[target_col]
X_test,  y_test  = test[features],  test[target_col]

print("Train/Valid/Test:", X_train.shape, X_valid.shape, X_test.shape)

# --- Baselines ---
# 1) Naive: y_hat_t = y_{t-1}
y_valid_naive = valid[f"{target_col}_lag1"]
rmse_naive = mean_squared_error(y_valid, y_valid_naive, squared=False)

# 2) Media móvil de 24h (ya creada como ma) — puede tener NaN iniciales, drop alineado
mask = ~valid[f"{target_col}_ma24"].isna()
rmse_ma24 = mean_squared_error(y_valid[mask], valid.loc[mask, f"{target_col}_ma24"], squared=False)

print(f"RMSE baseline naive (valid): {rmse_naive:.3f}")
print(f"RMSE baseline MA24 (valid):  {rmse_ma24:.3f}")
```

## 4) Modelado

### 4.1 LinearRegression (pipeline con escalado)
- Las lags/rolling pueden tener distintas escalas → **StandardScaler**.
- Validación con RMSE en `valid`.

### 4.2 RandomForestRegressor (búsqueda simple)
- `TimeSeriesSplit` para respetar el orden temporal.
- Pequeña malla de hiperparámetros para no alargar el *runtime*.


```
# --- Linear Regression ---
lin_pipe = Pipeline([
    ("scaler", StandardScaler(with_mean=True, with_std=True)),
    ("model", LinearRegression())
])
lin_pipe.fit(X_train, y_train)
pred_valid_lin = lin_pipe.predict(X_valid)
rmse_lin = mean_squared_error(y_valid, pred_valid_lin, squared=False)
print(f"RMSE LinearRegression (valid): {rmse_lin:.3f}")

# --- Random Forest con TSS ---
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
tscv = TimeSeriesSplit(n_splits=3)
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(random_state=42, n_jobs=-1)

param_grid = {
    "n_estimators": [200, 400],
    "max_depth": [6, 10, None],
    "min_samples_split": [2, 5]
}

gcv = GridSearchCV(rf, param_grid, scoring="neg_root_mean_squared_error", cv=tscv, n_jobs=-1, verbose=0)
gcv.fit(X_train, y_train)

best_rf = gcv.best_estimator_
pred_valid_rf = best_rf.predict(X_valid)
rmse_rf = mean_squared_error(y_valid, pred_valid_rf, squared=False)

print("Mejores hiperparámetros RF:", gcv.best_params_)
print(f"RMSE RandomForest (valid): {rmse_rf:.3f}")
```

## 5) Selección, evaluación en Test y *forecast* 24h

- Elegimos el mejor modelo en **valid**.  
- Re-entrenamos en **train+valid** y evaluamos en **test**.  
- Forecast **multi‑paso** (24h) de forma recursiva: usamos cada predicción como *lag* para la siguiente.


```
# --- Selección de modelo por RMSE en valid ---
valid_scores = {}
best_name = None
best_rmse = float("inf")

cands = [("LinearRegression", rmse_lin, lin_pipe),
         ("RandomForest", rmse_rf, best_rf)]

for name, rmse, model in cands:
    if rmse < best_rmse:
        best_rmse = rmse
        best_name = name
        best_model = model

print(f"Mejor en valid: {best_name} (RMSE={best_rmse:.3f})")

# --- Re-entrenar en train+valid ---
X_trv = pd.concat([X_train, X_valid], axis=0)
y_trv = pd.concat([y_train, y_valid], axis=0)

if best_name == "LinearRegression":
    final_model = Pipeline([("scaler", StandardScaler(with_mean=True, with_std=True)),
                            ("model", LinearRegression())])
    final_model.fit(X_trv, y_trv)
else:
    final_model = best_rf
    final_model.fit(X_trv, y_trv)

# --- Evaluación en Test ---
pred_test = final_model.predict(X_test)
rmse_test = mean_squared_error(y_test, pred_test, squared=False)
print(f"RMSE en TEST: {rmse_test:.3f}")

# --- Residuales simples ---
resid = y_test - pred_test
plt.figure(figsize=(10,3))
plt.plot(y_test.index, resid)
plt.title("Residuales en Test")
plt.xlabel("Tiempo"); plt.ylabel("Error")
plt.show()

# --- Forecast multi-paso (24h) ---
def make_future_features(df_ts, start_ts, periods, freq, last_features_df, target):
    '''
    Genera features para un calendario futuro recursivamente, usando predicciones
    como lags, y manteniendo rolling windows con un desplazamiento conceptual.
    '''
    future_idx = pd.date_range(start_ts, periods=periods, freq=freq)
    hist = df_ts.copy()
    feat_hist = last_features_df.copy()

    preds = []
    rows = []

    for ts in future_idx:
        base = {}
        # lags
        for l in [1,2,3,24]:
            base[f"{target}_lag{l}"] = hist[target].shift(l).iloc[-1]
        # rolling means/std (con shift(1) interpretado)
        for w in [3,24,168]:
            base[f"{target}_ma{w}"] = hist[target].shift(1).rolling(w).mean().iloc[-1]
            base[f"{target}_std{w}"] = hist[target].shift(1).rolling(w).std().iloc[-1]
        base["hour"] = ts.hour
        base["dow"] = ts.dayofweek
        base["is_weekend"] = int(base["dow"] >= 5)

        rows.append(base)

        X_row = pd.DataFrame([base], index=[ts])[feat_hist.drop(columns=[target]).columns]
        y_hat = final_model.predict(X_row)[0]
        preds.append(y_hat)

        # actualizar 'hist' con la predicción
        new_row = pd.DataFrame({target: [y_hat]}, index=[ts])
        hist = pd.concat([hist, new_row])

    future_df = pd.DataFrame(rows, index=future_idx)
    future_df[target] = preds
    return future_df

start = df_res.index.max() + pd.Timedelta(hours=1)
future_df = make_future_features(df_res[[target_col]].copy(), start, periods=24, freq="H",
                                 last_features_df=df_feat, target=target_col)

print("Forecast 24h (head):")
print(future_df[[target_col]].head())

out_path = Path("/mnt/data/forecast_24h.csv")
future_df[[target_col]].rename(columns={target_col:"prediction"}).to_csv(out_path, index_label="timestamp")
print("Archivo guardado:", out_path)
```

## 6) Checklist anti‑fuga y mejoras

**Anti‑fuga**  
- No usar valores del futuro para construir features del pasado (siempre `shift(1)` antes de *rolling*).  
- *Split* por tiempo; evitar `KFold` aleatorio.  
- No usar `scaler.fit` en todo el dataset antes del split.

**Mejoras posibles**  
- Afinar ventanas y lags con `TimeSeriesSplit`.  
- Probar `XGBoost/CatBoost/LightGBM` (si están disponibles).  
- Modelos SARIMAX/Prophet si el patrón es fuertemente estacional.  
- `Exogenous` variables: clima, eventos, etc.
