# Uso de Git: Introducción y Comandos Básicos

Este notebook tiene como objetivo explicar los conceptos básicos y comandos más utilizados de Git, un sistema de control de versiones distribuido. Aprenderás a iniciar un repositorio, hacer commits, trabajar con ramas y conectar con repositorios remotos.

## ¿Qué es Git?

Git es una herramienta de control de versiones que permite llevar un registro de los cambios en los archivos de un proyecto, facilitando la colaboración entre varios desarrolladores y permitiendo regresar a versiones anteriores del código en caso de ser necesario.

## Instalación de Git

Para utilizar Git, primero debes instalarlo. Puedes descargarlo desde la [página oficial de Git](https://git-scm.com/). Una vez instalado, abre una terminal y verifica la instalación con el siguiente comando:

````python
!git --version
````

## Inicialización de un Repositorio

Para comenzar a utilizar Git en tu proyecto, primero debes inicializar un repositorio. Ejecuta el siguiente comando en la terminal dentro del directorio de tu proyecto:

````python
!git init
````

## Agregar Cambios y Hacer un Commit

Una vez que hayas realizado cambios en tus archivos, puedes agregar estos cambios al área de staging y luego confirmar (commit) los cambios en el repositorio.

1. **Agregar archivos:** El comando `git add` se utiliza para agregar archivos específicos o todos los cambios al área de staging.
2. **Hacer commit:** El comando `git commit` registra los cambios en el repositorio junto con un mensaje descriptivo.

Ejemplo:

````python
!git add .
!git commit -m "Primer commit: se agregan los archivos iniciales"
````

## Uso de Ramas

Las ramas en Git te permiten trabajar en diferentes versiones del proyecto de forma paralela. Por ejemplo, puedes trabajar en una nueva funcionalidad sin afectar la rama principal (`master` o `main`).

Algunos comandos útiles:

- `git branch`: Muestra las ramas existentes.
- `git branch <nombre_rama>`: Crea una nueva rama.
- `git checkout <nombre_rama>`: Cambia a la rama especificada.

Ejemplo:

````python
!git branch nueva-funcionalidad
!git checkout nueva-funcionalidad
````

## Conectar con un Repositorio Remoto

Para colaborar con otros o tener una copia de seguridad en línea, puedes conectar tu repositorio local con un repositorio remoto (por ejemplo, en GitHub, GitLab o Bitbucket).

Los pasos básicos son:

1. **Agregar el remoto:**
   ```
   git remote add origin https://github.com/usuario/nombre-del-repositorio.git
   ```

2. **Enviar tus cambios al remoto:**
   ```
   git push -u origin master
   ```

Ejemplo:

````python
!git remote add origin https://github.com/usuario/nombre-del-repositorio.git
!git push -u origin master
````
