# Tutorial de Comandos CLI en Jupyter Notebook

En este notebook aprenderás a utilizar diversos comandos de la terminal (CLI) directamente desde Jupyter Notebook. Se presentan descripciones y ejemplos prácticos que van aumentando en complejidad, además de ejercicios propuestos para poner en práctica lo aprendido.

## Introducción

En este tutorial, exploraremos comandos básicos y avanzados de la línea de comandos. Muchos de estos comandos se pueden ejecutar utilizando el prefijo `!` en una celda de código, lo que permite ejecutar instrucciones del shell directamente desde Jupyter Notebook.

## 1. Comandos Básicos

A continuación se muestran ejemplos de comandos fundamentales para la navegación y manipulación de archivos.

````python
# Ejemplo de ls: Lista el contenido del directorio actual
!ls
````

````python
# Ejemplo de ls -r: Lista el contenido en orden inverso
!ls -r
````

````python
# Ejemplo de ls --help: Muestra la ayuda del comando ls
!ls --help
````

````python
# Ejemplo de man ls: Muestra el manual del comando ls
# Nota: En algunos entornos el comando 'man' puede no funcionar correctamente en notebooks.
!man ls
````

````python
# Ejemplo de pwd: Muestra el directorio actual de trabajo
!pwd
````

````python
# Ejemplo de cd: Cambiar de directorio
# En notebooks se recomienda usar el comando mágico %cd para cambiar de directorio
%cd ..
!pwd  # Verificar el cambio de directorio
````

````python
# Ejemplo de clear: Limpia la pantalla de la terminal
# Nota: Este comando puede no funcionar de la misma manera en notebooks.
!clear
````

````python
# Ejemplo de mkdir: Crea un nuevo directorio
!mkdir ejemplo_directorio
# Listamos para verificar que el directorio fue creado
!ls -l
````

````python
# Ejemplo de rm: Elimina un archivo
# Primero, creamos un archivo de prueba
!echo "Contenido de prueba" > archivo_prueba.txt
# Mostramos el archivo
!cat archivo_prueba.txt
# Eliminamos el archivo
!rm archivo_prueba.txt
# Verificamos que ya no existe
!ls
````

````python
# Ejemplo de rm -d -v: Elimina un directorio vacío de forma verbosa
!mkdir dir_vacio
!rm -d -v dir_vacio
````

````python
# Ejemplo de rm -d -i: Elimina un directorio vacío de forma interactiva
!mkdir dir_interactivo
# Se solicitará confirmación para eliminar el directorio
!rm -d -i dir_interactivo
````

````python
# Ejemplo de echo: Muestra un mensaje en la terminal
!echo "Hola, este es un mensaje de ejemplo"
````

````python
# Ejemplo de echo -e: Interpreta secuencias de escape (como \n para nueva línea)
!echo -e "Primera línea\nSegunda línea"
````

````python
# Ejemplo de cat: Muestra el contenido de un archivo
# Primero, creamos un archivo con contenido
!echo "Esta es la primera línea" > archivo_texto.txt
!echo "Esta es la segunda línea" >> archivo_texto.txt
# Mostramos el contenido del archivo
!cat archivo_texto.txt
````

````python
# Ejemplo de touch: Crea un archivo vacío (o actualiza su fecha de modificación)
!touch archivo_vacio.txt
# Verificamos la existencia del archivo
!ls -l archivo_vacio.txt
````

````python
# Ejemplo de mv: Mueve o renombra un archivo
# Renombramos 'archivo_vacio.txt' a 'archivo_renombrado.txt'
!mv archivo_vacio.txt archivo_renombrado.txt
# Verificamos el cambio
!ls -l archivo_renombrado.txt
````

````python
# Ejemplo de cp: Copia un archivo
# Copiamos 'archivo_texto.txt' a 'archivo_copia.txt'
!cp archivo_texto.txt archivo_copia.txt
# Verificamos el contenido de la copia
!cat archivo_copia.txt
````

````python
# Ejemplo de uso de carácter comodín '*'
# Creamos algunos archivos de ejemplo
!touch a_prueba_o.py
!touch a_test_o.py
!touch otro_archivo.py
# Listamos archivos que coincidan con el patrón a*o.py
!ls a*o.py
# Eliminamos archivos que coincidan con el patrón
!rm a*o.py
# Verificamos que fueron eliminados
!ls *.py
````

````python
# Ejemplo de uso de carácter comodín '[]'
# Creamos archivos de ejemplo
!touch ere.txt
!touch eye.txt
!touch eje.txt
# Listamos archivos que coincidan con el patrón e[ry]e.txt (busca 'ere.txt' o 'eye.txt')
!ls e[ry]e.txt
# Eliminamos archivos de prueba para limpiar
!rm ere.txt eye.txt eje.txt
````

````python
# Ejemplo de redirección con cat >
# Se utiliza para escribir contenido en un archivo
!echo "Contenido para file2display.txt" > file2display.txt
# Redireccionamos el contenido de 'file2display.txt' a 'another_text_file.txt'
!cat file2display.txt > another_text_file.txt
# Mostramos el contenido del nuevo archivo
!cat another_text_file.txt
````

````python
# Ejemplo de wc: Cuenta líneas, palabras y caracteres en un archivo
!wc file2display.txt
````

````python
# Ejemplo de head -n: Muestra las primeras n líneas de un archivo
# Creamos un archivo con varias líneas
!printf "Línea 1\nLínea 2\nLínea 3\nLínea 4\n" > multi_lineas.txt
!head -n 2 multi_lineas.txt
````

````python
# Ejemplo de tail -n: Muestra las últimas n líneas de un archivo
!tail -n 2 multi_lineas.txt
````

````python
# Ejemplo de find . -name: Busca archivos por nombre a partir del directorio actual
# Buscamos archivos con extensión .txt
!find . -name "*.txt"
````

````python
# Ejemplo de find . -type d -iname: Busca directorios (ignora mayúsculas/minúsculas) que coincidan con el patrón
!find . -type d -iname "ejemplo*"
````

````python
# Ejemplo de find . -mtime -2: Busca archivos modificados en los últimos 2 días
!find . -mtime -2
````

````python
# Ejemplo de grep -w: Busca coincidencias de palabra completa en un archivo
# Creamos un archivo con contenido de ejemplo
!echo "data information\ndata-analysis\ndata" > data.txt
!grep -w "data" data.txt
````

````python
# Ejemplo de grep -iw: Busca coincidencias ignorando mayúsculas/minúsculas
!echo "Data Information\ndata-analysis\ndata" > data2.txt
!grep -iw "data" data2.txt
````

````python
# Ejemplo de alias: Crear un alias para un comando
# Nota: Los alias creados de esta forma solo duran en la sesión actual.
!alias ll='ls -la'
# Uso del alias
!ll
````

````python
# Ejemplo de pipes: Usar el resultado de un comando como entrada para otro comando
# Pipe para filtrar la salida de ls que contenga la palabra "data"
!ls | grep "data"

# Pipe para mostrar la última línea de la salida de ls
!ls | tail -1
````

````python
# Ejemplo de history: Muestra el historial de comandos (puede variar según la configuración del kernel)
!history
````

## Ejercicio de Clase: Organización de Archivos

En este ejercicio se propone automatizar la organización de archivos en un directorio. Realiza lo siguiente:

1. Crea un directorio llamado `Proyecto`.
2. Dentro de `Proyecto`, crea tres archivos de texto: `archivo1.txt`, `archivo2.txt` y `archivo3.txt`.
3. Escribe en cada archivo un número diferente de líneas:
   - `archivo1.txt`: 2 líneas
   - `archivo2.txt`: 3 líneas
   - `archivo3.txt`: 1 línea
4. Utiliza `wc -l` para contar las líneas de cada archivo y determinar cuál tiene menos líneas.
5. Crea un subdirectorio llamado `Resumen` dentro de `Proyecto` y mueve el archivo con menos líneas a dicho subdirectorio.
6. Finalmente, muestra el contenido del archivo movido.

````python
# Solución al ejercicio de organización de archivos

# 1. Crear el directorio 'Proyecto'
!mkdir -p Proyecto

# 2. Crear tres archivos de texto dentro de 'Proyecto'
!echo -e "Línea 1\nLínea 2" > Proyecto/archivo1.txt
!echo -e "Línea 1\nLínea 2\nLínea 3" > Proyecto/archivo2.txt
!echo "Línea 1" > Proyecto/archivo3.txt

# 3. Contar las líneas de cada archivo
!wc -l Proyecto/archivo*.txt

# 4. Identificar el archivo con menos líneas (en este caso, 'archivo3.txt')
# 5. Crear el subdirectorio 'Resumen' y mover el archivo con menos líneas
!mkdir -p Proyecto/Resumen
!mv Proyecto/archivo3.txt Proyecto/Resumen/

# 6. Mostrar el contenido del archivo movido
!cat Proyecto/Resumen/archivo3.txt
````

## Ejercicio Adicional: Búsqueda y Filtrado

En este ejercicio se utilizarán comandos de búsqueda y filtrado para procesar un archivo de registro.

1. Crea un archivo llamado `registro.txt` con varias líneas de texto que contengan diferentes palabras (por ejemplo, "error", "aviso", "info").
2. Utiliza `grep` para extraer solo las líneas que contienen la palabra "error" (sin importar mayúsculas o minúsculas) y redirige la salida a un nuevo archivo llamado `errores.txt`.
3. Finalmente, utiliza `wc -l` para contar cuántas líneas tiene el archivo `errores.txt`.

````python
# Solución al ejercicio de búsqueda y filtrado

# 1. Crear el archivo 'registro.txt' con varias líneas
!echo -e "Error: fallo en el módulo A\nAviso: uso de memoria alto\nInfo: proceso completado\nerror: fallo en el módulo B\nINFO: inicio de sesión exitoso" > registro.txt

# 2. Extraer líneas que contengan 'error' (ignorando mayúsculas/minúsculas) y redirigir a 'errores.txt'
!grep -i "error" registro.txt > errores.txt

# 3. Contar las líneas de 'errores.txt'
!wc -l errores.txt

# Mostrar el contenido de 'errores.txt'
!cat errores.txt
````

## 2. Instalación y Configuración de Herramientas

En esta sección se abordan aspectos relacionados con la instalación y configuración de herramientas de desarrollo.

### Editor de Código: Visual Studio Code (VSCode)

Para instalar Visual Studio Code, sigue estos pasos:
1. Descarga el instalador desde la página oficial: [VSCode Download](https://code.visualstudio.com/).
2. Ejecuta el instalador y sigue las instrucciones.
3. Una vez instalado, puedes abrir VSCode desde la terminal usando el comando `code`.

_Ejemplo:_
```bash
code .
```

### Instalación de Anaconda

Anaconda es una distribución de Python que incluye numerosos paquetes y herramientas para ciencia de datos.

Pasos básicos para instalar Anaconda:
1. Descarga el instalador desde [Anaconda Download](https://www.anaconda.com/products/distribution).
2. Ejecuta el instalador y sigue las instrucciones.
3. Una vez instalado, puedes gestionar entornos y paquetes con `conda`.

_Ejemplo para crear un entorno:_
```bash
conda create -n mi_entorno python=3.8
```

### Instalación de Paquetes con Conda y Pip

Para instalar paquetes, puedes usar `conda` o `pip`.

_Ejemplos:_

- Con conda:
```bash
conda install numpy pandas
```

- Con pip:
```bash
pip install numpy pandas
```
