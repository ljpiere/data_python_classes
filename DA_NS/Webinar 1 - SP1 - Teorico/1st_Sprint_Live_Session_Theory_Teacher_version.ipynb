Example 6. Max sent a message to John, but the message came with # in between the words and spaces at both the beginning and the end of the message and with different letter cases. Example 8. Later that day, John sent a list of good bitcoin books to Max. He is a nice friend, so he sent it in the form of a python list and even added the authors at the end of it. But there's one book with more than one author, and Max wants to split the authors and store them separately in the form of a list. Find the book that has more than 1 author, extract it using indexing, split the Authors' names into a list and print the result.

```python
bitcoin_info = [
    ['Mastering Bitcoin: Unlocking Digital Cryptocurrencies', 'USA', 2014, ['cryptocurrency', 'technology'], 'Andreas M. Antonopoulos'],
    ['The Age of Cryptocurrency: How Bitcoin and Digital Money Are Challenging the Global Economic Order', 'USA', 2015, ['cryptocurrency', 'economics'], 'Paul Vigna, Michael J. Casey'],
    ['Digital Gold: Bitcoin and the Inside Story of the Misfits and Millionaires Trying to Reinvent Money', 'USA', 2015, ['cryptocurrency', 'finance'], 'Nathaniel Popper'],
    ['The Bitcoin Standard: The Decentralized Alternative to Central Banking', 'USA', 2018, ['cryptocurrency', 'economics'], 'Saifedean Ammous'],
    ['Bitcoin Billionaires: A True Story of Genius, Betrayal, and Redemption', 'USA', 2019, ['cryptocurrency', 'biography'], 'Ben Mezrich']
]

authors = bitcoin_info[1][-1]

authors_split = authors.split(',')
print(authors_split)
```
