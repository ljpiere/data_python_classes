**Pista:** Define una función personalizada para la condición.

````python
def categoria_piso(piso):
    etiqueta = ''
    if piso == 1:
        etiqueta = 'primer piso'
    elif piso == 2:
        etiqueta = 'segundo piso'
    elif piso == 3:
        etiqueta = 'tercer piso'
    else:
        etiqueta = 'más de 3 pisos'
    return etiqueta
````
