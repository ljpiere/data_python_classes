# EDA del dataset de Grammy Nominees & Winners (1958–2024)
En este notebook realizaremos pasos básicos de análisis exploratorio, filtrado, renombrado, manejo de nulos e `groupby` con funciones.

## 1. Importación de librerías y carga de datos

```python
import pandas as pd

# Cargar el dataset
df = pd.read_csv('https://raw.githubusercontent.com/ljpiere/data_python_classes/main/DA_DS/datasets/Grammy_Nominees_Winners_1958to2024.csv')
df.head()
```

## 2. EDA básico (sin gráficas)
- `info()` para tipos y no nulos
- `describe()` para estadísticas descriptivas
- `isnull().sum()` para conteo de nulos

```python
df.info()

df.describe(include='all')

df.isnull().sum()
```

## 3. Filtros básicos
- Filtrar ganadores (Winner == True)
- Filtrar por categoría específica
- Filtrar por año

```python
# Ganadores
winners = df[df['Winner'] == True]
winners.head()

# Nominees de 'Record of the Year'
record_nominees = df[df['Category'] == 'Record of the Year']
record_nominees.head()

# Entradas de 2024
year_2024 = df[df['Year'] == 2024]
year_2024
```

## 4. Cambios de nombres de columna
Renombrar columnas a snake_case para estandarizar.

```python
# Usando bucle for y métodos string
for col in df.columns:
    new_col = col.strip().lower().replace(' ', '_').replace('/', '_')
    df.rename(columns={col: new_col}, inplace=True)
df.columns
```

## 5. Búsqueda e imputación de vacíos y nulos
- Identificar columnas con nulos
- Imputar valores faltantes

```python
# Conteo de nulos por columna
df.isnull().sum()

# Imputación: rellenar valores faltantes hacia adelante
df_ffill = df.fillna(method='ffill')
df_ffill.isnull().sum()

# Imputación con valor fijo en 'Nominee'
if 'nominee' in df_ffill.columns:
    df_ffill['nominee'].fillna('Unknown', inplace=True)
df_ffill['nominee'].isnull().sum()
```

## 6. `groupby` con funciones
Definiremos funciones que encapsulan operaciones de agrupación.

### 6.1 Conteo de ganadores por año
Función: cuenta cuántos ganadores hubo cada año.

```python
def count_winners_by_year(df):
    return df[df['winner'] == True].groupby('year')['winner'].count()

count_winners_by_year(df_ffill).head()
```

### 6.2 Promedio de nominaciones por categoría
Función: calcula el número promedio de nominaciones por categoría.

```python
def avg_nominees_per_category(df):
    counts = df.groupby('category')['nominee'].count()
    return counts.mean()

avg_nominees_per_category(df_ffill)
```
