## 1. Variables y `print()`
**Ejercicio Avanzado:**
```python
lat = 4.7109886
lon = -74.072092
city = 'Bogota'
temp = 23.67
```
- Formatea e imprime:
`La ciudad de Bogotá (4.711, -74.072) tiene una temperatura de 23.7°C.`


````python
# Solución Ejercicio 1
lat = 4.7109886
lon = -74.072092
city = 'Bogota'
temp = 23.67
msg = f"La ciudad de {city} ({lat:.3f}, {lon:.3f}) tiene una temperatura de {temp:.1f}°C."
print(msg)
````

## 2. Manejo Avanzado de Cadenas
**Ejercicio Avanzado:**
```python
data = 'prod1:20|prod2:35|prod3:12|prod4:50'
```
- Separa manualmente cada registro sin usar estructuras de mapeo ni condicionales.
- Crea variables numéricas a partir de cada valor.
- Asigna a `high_stock` la lista de productos con stock manualmente identificados (>15).


````python
# Solución Ejercicio 2
data = 'prod1:20|prod2:35|prod3:12|prod4:50'
# Separar sin bucles ni mapeo
d0, d1, d2, d3 = data.split('|')
# Extraer valores numéricos manualmente
p0, v0 = d0.split(':'); p1, v1 = d1.split(':')
p2, v2 = d2.split(':'); p3, v3 = d3.split(':')
stock0 = int(v0)
stock1 = int(v1)
stock2 = int(v2)
stock3 = int(v3)
# Identificar manualmente los que tienen stock >15
high_stock = [p1, p3]
print('Stocks:', stock0, stock1, stock2, stock3)
print('Alta demanda:', high_stock)
````

## 3. Conversión de Tipos Complejos
**Ejercicio Avanzado:**
```python
mixta = ['100', '200.5', '300', 'invalid', '400']
```
- Convierte manualmente cada elemento, asigna `None` si falla.


````python
# Solución Ejercicio 3
mixta = ['100', '200.5', '300', 'invalid', '400']
# Conversión manual sin estructuras de control
try:
    conv0 = int(mixta[0])
except ValueError:
    conv0 = None
try:
    conv1 = float(mixta[1])
except ValueError:
    conv1 = None
try:
    conv2 = int(mixta[2])
except ValueError:
    conv2 = None
try:
    conv3 = int(mixta[3])
except ValueError:
    conv3 = None
try:
    conv4 = int(mixta[4])
except ValueError:
    conv4 = None
convertidos = [conv0, conv1, conv2, conv3, conv4]
print(convertidos)
````

## 4. Gestión Avanzada de Excepciones
**Ejercicio Avanzado:**
```python
vals = [10, 'a', 20]
```
- Calcula la media sumando los elementos manualmente y maneja errores específicos.


````python
# Solución Ejercicio 4
vals = [10, 'a', 20]
try:
    total = vals[0] + vals[1] + vals[2]
    media = total / 3
    print(f"Media: {media}")
except ZeroDivisionError:
    print("Error: lista vacía.")
except TypeError:
    print("Error: tipos incompatibles en la lista.")
````

## 5. Listas e Indexación Anidada
**Ejercicio Avanzado:**
```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
```
- Extrae la diagonal principal y la última columna sin bucles.


````python
# Solución Ejercicio 5
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
d0 = matrix[0][0]; d1 = matrix[1][1]; d2 = matrix[2][2]
diagonal = [d0, d1, d2]
c0 = matrix[0][2]; c1 = matrix[1][2]; c2 = matrix[2][2]
last_column = [c0, c1, c2]
print('Diagonal:', diagonal)
print('Última columna:', last_column)
````

## 6. Formateo Avanzado de Cadenas en Listas
**Ejercicio Avanzado:**
```python
ventas = [1234.5, 2345.67, 3456.78]
meses = ['Ene', 'Feb', 'Mar']
```
- Crea los mensajes manualmente sin bucles.


````python
# Solución Ejercicio 6
ventas = [1234.5, 2345.67, 3456.78]
meses = ['Ene', 'Feb', 'Mar']
m1 = f"En {meses[0]} se vendió ${ventas[0]:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.')
m2 = f"En {meses[1]} se vendió ${ventas[1]:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.')
m3 = f"En {meses[2]} se vendió ${ventas[2]:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.')
report = [m1, m2, m3]
print(report)
````

## 7. Funciones Integradas (`len`, `append`, `count`)
**Ejercicio Avanzado:**
```python
items = ['a', 'b', 'a', 'c']
```
- Obtén la frecuencia de 'a', añade elementos y revisa cambios sin condicionales.


````python
# Solución Ejercicio 7
items = ['a', 'b', 'a', 'c']
freq0 = items.count('a')
items.append('d')
len1 = len(items)
items.append('a')
freq1 = items.count('a')
print(f"Frecuencia inicial: {freq0}")
print(f"Longitud tras append: {len1}")
print(f"Frecuencia tras nuevo append: {freq1}")
````
