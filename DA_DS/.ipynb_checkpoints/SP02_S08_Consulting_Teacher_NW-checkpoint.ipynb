## 1. Análisis de disponibilidad crítica

### Ejercicio 1.1: Promedio de stock bajo
**Enunciado:**  
Crea una función `promedio_stock_bajo(inventario, umbral)` que calcule el **promedio de disponibilidad** de los productos cuyo stock está por debajo del umbral.

**Pista:**  
Suma todos los productos con disponibilidad < umbral y divídelos entre la cantidad de casos encontrados.


````python
def promedio_stock_bajo(inventario, umbral):
    total = 0
    cantidad = 0
    for producto in inventario:
        if producto[4] < umbral:
            total += producto[4]
            cantidad += 1
    if cantidad == 0:
        return 0
    return total / cantidad
````

## 2. Simulación de combos promocionales

### Ejercicio 2.1: Crear combo
**Enunciado:**  
Define una función `crear_combo(inventario, lista_ids)` que calcule el **total con descuento del 10%** de un grupo de productos por ID.

**Pista:**  
Suma el precio de los productos encontrados en la lista y aplica el descuento al final.


````python
def crear_combo(inventario, lista_ids):
    total = 0
    for producto in inventario:
        if producto[0] in lista_ids:
            total += producto[8]
    descuento = total * 0.10
    return total - descuento
````

## 3. Revisión de duplicados en listas

### Ejercicio 3.1: Detección de IDs repetidos
**Enunciado:**  
Escribe una función `detectar_repetidos(lista_ids)` que retorne una lista con los elementos repetidos.

**Pista:**  
Usa un diccionario o una lista auxiliar para contar.


````python
def detectar_repetidos(lista_ids):
    vistos = []
    repetidos = []
    for id in lista_ids:
        if id in vistos and id not in repetidos:
            repetidos.append(id)
        else:
            vistos.append(id)
    return repetidos
````

## 5. Simulación de inventario dinámico

### Ejercicio 5.1: Restock automático
**Enunciado:**  
Crea una función `reponer_inventario(inventario, minimo, cantidad_reposicion)` que recorra el inventario y aumente la disponibilidad de todos los productos con stock menor al mínimo indicado.

**Pista:**  
Modifica directamente `producto[4]` si es menor a `minimo`.


````python
def reponer_inventario(inventario, minimo, cantidad_reposicion):
    for producto in inventario:
        if producto[4] < minimo:
            producto[4] += cantidad_reposicion
            print(f"{producto[2]} repuesto a {producto[4]} {producto[6]}")
````

## 6. Reporte completo de inventario

### Ejercicio 6.1: Generar reporte tabulado
**Enunciado:**  
Define una función `generar_reporte(inventario)` que imprima una tabla con encabezados:

```
ID | Nombre | Stock | Unidad | Precio
```

Cada fila debe mostrar los datos del producto alineados correctamente.

**Pista:**  
Usa `str.ljust()` o `str.rjust()` para alinear.


````python
def generar_reporte(inventario):
    print(f"{'ID':<5} {'Nombre':<20} {'Stock':<8} {'Unidad':<10} {'Precio':<8}")
    print("-" * 55)
    for producto in inventario:
        print(f"{producto[0]:<5} {producto[2]:<20} {producto[4]:<8} {producto[6]:<10} ${producto[8]:<8.2f}")
````
