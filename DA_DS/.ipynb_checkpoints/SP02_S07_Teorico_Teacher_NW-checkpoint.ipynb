## 1. Definición de Funciones
### Ejercicio 1.1: Función de saludo (Básico)
**Caso de uso:** Queremos una función que salude a un usuario por su nombre.
- Define `def saludar(nombre):` que retorne "Hola, {nombre}!".

**Pista:** Usa un f-string y `return`.

````python
# Solución Ejercicio 1.1
def saludar(nombre):
    crear_saludo = f"Hola, {nombre}!"
    return crear_saludo

variable1 = 45
saludar(variable1)
````

### Ejercicio 1.2: Cálculo de área (Intermedio)
**Caso de uso:** En una aplicación de diseño, necesitamos calcular áreas de rectángulos.
- Define `def area_rectangulo(base, altura):` que retorne el área.
- Si alguno de los parámetros es <= 0, retorna `None`.

**Pista:** Valida parámetros antes de calcular.

````python
# Solución Ejercicio 1.2
def area_rectangulo(base, altura):
    v1 = base*altura
    return v1

v1 = 3
v2 = 4
v3 = area_rectangulo(v1, v2)
print(v3)
````

## 2. Métodos de listas en bucles

### Ejercicio 2.1: Contar palabras con `a` (Básico)
**Caso de uso:** Contar cuántas palabras en una lista contienen la letra `"a"`.

**Pista:** Usa `in` con `palabra.lower()`.


````python
palabras = ["Casa", "Sol", "Agua", "Montaña"]
contador = 0
for palabra in palabras:
    if "a" in palabra.lower():
        contador += 1
print(f"Palabras con 'a': {contador}")
````

### Ejercicio 2.2: Normalizar entrada de correos (Intermedio)
**Caso de uso:** Convertir todos los correos en una lista a minúsculas y guardarlos en una nueva lista.

**Pista:** Usa `.lower()` y un bucle para poblar la nueva lista.


````python
correos = ["USUARIO@MAIL.com", "Admin@SERVIDOR.NET", "info@pagina.ORG"]
normalizados = []
for correo in correos:
    normalizados.append(correo.lower())

print(normalizados)
````

## 3. Operadores Lógicos en Funciones

Combinamos comparaciones con `and`, `or`, `==` para decisiones más complejas.

### Ejercicio 3.1: Rango válido (Básico)
**Caso de uso:** Verificamos si una edad está entre 18 y 65 para una encuesta.
- Define `def edad_valida(edad):` que retorne `True` si `18 <= edad <= 65`, `False` de lo contrario.

**Pista:** Usa `and`.

````python
# Solución Ejercicio 3.1
def edad_valida(edad):
    if edad >= 18 and edad <= 65:
        return True
    else:
        return False
````

````python
edad_valida(17)
````

### Ejercicio 3.2: Verificación de credenciales (Intermedio)
**Caso de uso:** En un login, aceptamos usuarios y contraseñas específicas.
- Define `def validar_usuario(user, pwd):`
  - Si `user == 'admin' and pwd == '1234'`, retorna 'Acceso concedido'.
  - Si `user == 'admin' and pwd != '1234'`, retorna 'Contraseña incorrecta'.
  - Si `user != 'admin'`, retorna 'Usuario no reconocido'.

**Pista:** Usa `and`, `or` y comparaciones.  

````python
# Solución Ejercicio 3.2
def validar_usuario(user, pwd):
    if user == 'admin' and pwd == '1234':
        return 'Acceso concedido'
    elif user == 'admin' and pwd != '1234':
        return 'Contraseña incorrecta'
    else:
        return 'Usuario no reconocido'

print(validar_usuario('admin','1234'))
print(validar_usuario('admin','0000'))
print(validar_usuario('juan','1234'))
````

## 4. Funciones de Complejidad Media (Sin Bucles)
### Ejercicio 4.1: Contar vocales en un texto (Intermedio)
**Caso de uso:** Analizamos reseñas de clientes y queremos contar cuántas vocales contienen.
- Define `def contar_vocales(texto):`
  - Convierte `texto` a minúsculas.
  - Recorre cada carácter y cuenta si es `a, e, i, o, u`.
  - Retorna el conteo.

**Pista:** Usa un bucle `for` interno y `in 'aeiou'`.

````python
# Solución Ejercicio 4.1
def contar_vocales(texto):
    t = texto.lower()
    count_a = t.count('a')
    count_e = t.count('e')
    count_i = t.count('i')
    count_o = t.count('o')
    count_u = t.count('u')
    return count_a + count_e + count_i + count_o + count_u

print(contar_vocales('Data Science'))
````

### Ejercicio 4.2: Aplicar descuento a lista de precios (Intermedio)
**Caso de uso:** Tenemos precios en una lista y queremos aplicar descuento:
```python
precios = [100, 250, 30, 75, 200]
```
- Define `def aplicar_descuento(precios, descuento_pct):`
  - Recorre `precios`, aplica el descuento porcentual y almacena los nuevos precios en una lista.
  - Retorna la lista resultante.

**Pista:** Usa un bucle `for` y `append`.  

````python
# Solución Ejercicio 4.2
def aplicar_descuento(precios, descuento_pct):
    # Asumimos 5 precios
    p0, p1, p2, p3, p4 = precios
    d0 = p0 * (1 - descuento_pct/100)
    d1 = p1 * (1 - descuento_pct/100)
    d2 = p2 * (1 - descuento_pct/100)
    d3 = p3 * (1 - descuento_pct/100)
    d4 = p4 * (1 - descuento_pct/100)
    result = []
    result.append(d0)
    result.append(d1)
    result.append(d2)
    result.append(d3)
    result.append(d4)
    return result

print(aplicar_descuento([100, 250, 30, 75, 200], 10))
````

## Most common mistakes

### Paso 1: Limpieza inicial de un registro de cliente
**Errores comunes:**
- No eliminar espacios en blanco: olvidar usar strip() y trabajar con cadenas con espacios.
- Acceso incorrecto por índice: confundir la posición de cada campo en la lista.
- No convertir tipos: tratar strings como números sin convertir con int().

**Ejercicio: Dado el registro [' Juan ', 'Doe', ' 30 '], limpia los espacios y convierte la edad a entero.**

### Paso 2: Convertir las categorías a minúsculas
**Errores comunes:**
- Aplicar .lower() a la lista en lugar de a cada string.
- No reasignar el resultado de lower(): `categoria.lower()` sin actualizar la lista.
- Ignorar mayúsculas y minúsculas en comparaciones posteriores.

**Ejercicio: Convierte la lista de categorías ['VENTAS', 'MARKETING', 'RRHH'] a versiones en minúsculas.**

### Paso 3: Limpieza completa de un usuario
**Errores comunes:**
- Modificar la lista original en lugar de trabajar con una copia.
- Escribir mal el nombre de la función o parámetros incorrectos.
- Problemas de indentación dentro del loop.

**Ejercicio: Implementa una función que reciba [' Ana ', ' 28 '] y devuelva una lista limpia con tipos correctos.**

### Paso 4: Limpieza de toda la base de usuarios
**Errores comunes:**
- Olvidar retornar la lista limpia al final de la función.
- Reasignar variables dentro del loop y perder datos.
- No manejar entradas vacías o None en los registros.

**Ejercicio: Aplica tu función de limpieza a la lista de usuarios [[' Juan ', 'Doe', '30'], ['Luisa','Perez','25']].**

### Paso 5: Calcular ingresos totales
**Errores comunes:**
- Usar sum() sobre un string en vez de sobre números.
- Acceder al índice equivocado para los valores de gasto.
- No inicializar la variable acumuladora antes del bucle.

**Ejercicio: Calcula el ingreso total de la lista de gastos [120.5, 89.99, 45] usando un bucle.**

### Paso 6: Usuarios menores de 30 años
**Errores comunes:**
- Usar `<= 30` en lugar de `< 30` según el criterio definido.
- Confundir operadores lógicos `and`/`or` al combinar condiciones.
- No comprobar el tipo de datos antes de la comparación.

**Ejercicio: Filtra los usuarios [{'name':'Ana','age':25}, {'name':'Luis','age':30}] para quienes age < 30.**

### Paso 7: Jóvenes con alto gasto
**Errores comunes:**
- No especificar correctamente la condición del filtro.
- Olvidar ordenar o seleccionar correctamente los resultados.
- No manejar el caso de lista vacía al aplicar max() o sort().

**Ejercicio: Entre los gastos [100, 200, 150], encuentra el gasto máximo para usuarios menores de 25 años.**

### Paso 8: Usuarios que compraron ropa
**Errores comunes:**
- Sensibilidad a mayúsculas: no normalizar strings en la comparación.
- Comparar con literales sin verificar espacios o caracteres especiales.
- No devolver una lista vacía cuando no hay coincidencias.

**Ejercicio: De la lista de compras ['alimentos','ropa','electrónicos'], devuelve solo los que sean 'ropa'.**

### Paso 9: Función con filtro por categoría
**Errores comunes:**
- No definir la función antes de invocarla.
- Olvidar el `return` dentro de la función de filtrado.
- No documentar el parámetro de categoría y su tipo esperado.

**Ejercicio: Define y usa una función filter_by_category('deporte', ['deporte','ocio','cultura']).**

### Paso 10: Buenas prácticas y depuración
**Errores comunes:**
- No añadir comentarios o docstrings para explicar el código.
- No incluir manejo de excepciones para inputs inesperados.
- No probar casos límite ni datos erróneos.

**Ejercicio: Agrega un docstring y manejo de excepción a una función que divide dos números.**
