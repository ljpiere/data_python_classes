## 1. Bucle `while`

### Ejercicio 1.1: Contador simple (Básico)
**Caso de uso:** Queremos imprimir los números del 1 al 5 usando un bucle `while`.

**Pista:** inicializa un contador y repite mientras sea menor o igual a 5.


```
contador = 1
while contador <= 5:
    print(contador)
    contador += 1
```

### Ejercicio 1.2: Validación de entrada (Intermedio)
**Caso de uso:** Solicita al usuario que ingrese la palabra "ok". Repite hasta que lo haga correctamente.

**Pista:** Usa `input()` dentro de un bucle `while`.


```
entrada = ""
while entrada.lower() != "ok":
    entrada = input("Escribe 'ok' para continuar: ")
print("¡Gracias!")
```

## 2. Bucle `for` simple

### Ejercicio 2.1: Iterar una lista (Básico)
**Caso de uso:** Recorre una lista de frutas e imprime cada una en mayúsculas.

**Pista:** Usa `for` con `fruta.upper()`.


```
frutas = ["manzana", "banana", "kiwi"]
for fruta in frutas:
    print(fruta.upper())
```

### Ejercicio 2.2: Contar elementos numéricos (Intermedio)
**Caso de uso:** Cuenta cuántos elementos en una lista son mayores a 10.

**Pista:** Usa un contador dentro del `for`.


```
numeros = [5, 12, 7, 18, 3, 25]
contador = 0
for n in numeros:
    if n > 10:
        contador += 1
print(f"Mayores a 10: {contador}")
```

## 3. Bucle `for` anidado

### Ejercicio 3.1: Tablas de multiplicar (Básico)
**Caso de uso:** Imprimir las tablas del 1 al 3, hasta el 5.

**Pista:** Usa dos bucles `for` anidados.


```
for i in range(1, 4):
    print(f"Tabla del {i}")
    for j in range(1, 6):
        print(f"{i} x {j} = {i*j}")
    print("----")
```

### Ejercicio 3.2: Buscar coincidencias en matrices (Intermedio)
**Caso de uso:** Dada una matriz de listas, cuenta cuántas veces aparece el número 0.

**Pista:** Recorre la matriz con bucles anidados y usa una variable acumuladora.


```
matriz = [
    [0, 1, 2],
    [4, 0, 6],
    [0, 8, 9]
]
conteo = 0
for fila in matriz:
    for valor in fila:
        if valor == 0:
            conteo += 1
print(f"Cantidad de ceros: {conteo}")
```

## 4. Métodos de listas en bucles

### Ejercicio 4.1: Contar palabras con `a` (Básico)
**Caso de uso:** Contar cuántas palabras en una lista contienen la letra `"a"`.

**Pista:** Usa `in` con `palabra.lower()`.


```
palabras = ["Casa", "Sol", "Agua", "Montaña"]
contador = 0
for palabra in palabras:
    if "a" in palabra.lower():
        contador += 1
print(f"Palabras con 'a': {contador}")
```

### Ejercicio 4.2: Normalizar entrada de correos (Intermedio)
**Caso de uso:** Convertir todos los correos en una lista a minúsculas y guardarlos en una nueva lista.

**Pista:** Usa `.lower()` y un bucle para poblar la nueva lista.


```
correos = ["USUARIO@MAIL.com", "Admin@SERVIDOR.NET", "info@pagina.ORG"]
normalizados = []
for correo in correos:
    normalizados.append(correo.lower())

print(normalizados)
```
