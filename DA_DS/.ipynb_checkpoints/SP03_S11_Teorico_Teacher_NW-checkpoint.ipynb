## Parte 2: Renombrar columnas
Automatizar cambios de nombres es clave para estandarizar datasets.

### Método: bucle `for` con `rename()`
- Iteramos sobre `df.columns`.
- **Uso**: `df.rename(columns={old: new}, inplace=True)`.
- **Ejercicio 2.1**: Convierte a snake_case y elimina caracteres especiales.

```python
nuevos_nombres = []
for col in df_house.columns:
    new = col.strip().lower().replace(' ', '_').replace('-', '_')
    nuevos_nombres.append(new)

df_house.columns = nuevos_nombres

df_house.columns
```

```python
for col in df_house.columns:
    new = col.strip().lower().replace(' ', '_').replace('-', '_')
    df_house.rename(columns={col: new}, inplace=True)
df_house.columns
```

### Método: `map()` con función propia
- Definir función `clean(col)` y aplicar a `df.columns.map(clean)`.
- **Ventaja**: Reutilización entre DataFrames.
- **Ejercicio 2.2**: Aplica tu función a `df_divisa`.

```python
def clean(col):
    return (col.strip()
               .lower()
               .replace(' ', '_')
               .replace('/', '_per_'))
df_divisa.columns = df_divisa.columns.map(clean)
df_divisa.columns
```

## Parte 3: Filtrado y selección
Existen varias formas de extraer subconjuntos de datos.

### Método: Boolean indexing
- Selecciona filas que cumplen condiciones.
- **Ejemplo**: `df[df['price'] > 500000]`.
- **Ejercicio 3.1**: Obtén casas con `bedrooms` >= 3 y `price` < 300000.

```python
df_house[(df_house['habitaciones'] >= 3) & (df_house['precio_(eur)'] < 300000)]
```

### Método: `.loc[row_cond, col_list]`
- Filtra filas y columnas juntos.
- **Ejemplo**: `df.loc[cond, ['col1','col2']]`.
- **Ejercicio 3.2**: Selecciona `date`, `currency`, `rate` de `df_divisa` donde `rate` es el Top 5%.

```python
df_divisa.columns
```

```python
top5 = df_divisa['Volume'].quantile(0.95)

df_divisa.loc[df_divisa['Volume'] >= top5, ['ID', 'Currency Pair', 'Bid Price']]
```

### Método: `.iloc[pos_filas, pos_cols]`
- Selección por posición entera.
- **Ejercicio 3.3**: Muestra primeras 7 filas y últimas 4 columnas de `df_house`.

```python
df_house.iloc[:7, -4:]
```

## Parte 4: `groupby` y agregaciones avanzadas
Combina agrupación con funciones `agg` para estadísticas múltiples.

### Método: `groupby(...).agg({...})`
- Permite definir múltiples agregaciones por columna.
- **Ejercicio 4.1**: Para `df_house`, agrupa por `bedrooms` y calcula:
    - `mean` y `median` de `price`.
    - `count` de registros.

```python
df_house.columns
```

```python
df_house.groupby("ciudad")["baños"].median()
```

```python
df_house.groupby(['habitaciones', 'baños']).agg(
    price_mean=('precio_(eur)','mean'),
    price_median=('precio_(eur)','median'),
    total=('precio_(eur)','count')
)
```

### Método: `pivot_table()`
- Tablas dinámicas, similar a Excel.
- **Ejercicio 4.2**: Crea pivot table de `df_divisa` para promedio de `rate` por `currency` y año.

```python
df_divisa['year'] = df_divisa['date'].dt.year
pd.pivot_table(df_divisa, values='rate', index='currency', columns='year', aggfunc='mean')
```

## Parte 5: Gestión de valores faltantes (`fillna` y `dropna`)
Diferentes estrategias según el contexto de negocio.

```python
df_house.head()
```

### Método: Imputación con valor fijo o estadístico
- `df['col'].fillna(valor, inplace=True)`.
- **Ejercicio 5.1**: Imputa nulos de `df_house['sqft_living']` con la mediana.

```python
df_house['habitaciones'].fillna(df_house['habitaciones'].median(), inplace=True)
df_house['habitaciones'].isnull().sum()
```

### Método: forward/backward fill
- `method='ffill'` o `method='bfill'`.
- **Ejercicio 5.2**: Aplica ffill en `df_divisa` y compara con bfill.

```python
df_divisa['rate_ffill'] = df_divisa['rate'].fillna(method='ffill')
df_divisa['rate_bfill'] = df_divisa['rate'].fillna(method='bfill')
df_divisa[['rate','rate_ffill','rate_bfill']].head(8)
```

### Método: `dropna()`
- Elimina filas/columnas con nulos.
- **Ejercicio 5.3**: Elimina filas con más de 2 nulos en `df_house`. ¿Cuántas quedan?

```python
df_house.dropna(thresh=df_house.shape[1]-2).shape
```

## Parte 6: Duplicados (`duplicated` y `drop_duplicates`)
Cómo detectar y manejar registros repetidos.

### Método: `duplicated()`
- Marca filas duplicadas.
- **Ejercicio 6.1**: ¿Cuántos duplicados hay en `df_divisa` considerando `date` y `currency`?

```python
df_divisa.duplicated().sum()
```

```python
df_divisa.duplicated(subset=['ID','Currency Pair']).sum()
```

### Método: `drop_duplicates()`
- Elimina duplicados, eligiendo `keep`.
- **Ejercicio 6.2**: Crea `clean_divisa` sin duplicados y comprueba.

```python
clean_divisa = df_divisa.drop_duplicates(subset=['date','currency'], keep='first')
clean_divisa.shape
```

## Parte 7: Merge y mezcla de DataFrames
Uniendo información de ambos datasets.

### Ejercicio 7.1: `merge()`
- Une `df_house` con `df_divisa` por fecha: cómo convertir `df_house['date']` a datetime.
- Realiza un merge `left` y responde: ¿Cuántos registros sin correspondencia?

```python
df_house['date'] = pd.to_datetime(df_house['date'])  # si existe
merged = df_house.merge(df_divisa, on='date', how='left')
merged['rate'].isnull().sum()
```

### Ejercicio 7.2: `apply()` y funciones vectorizadas
- **Ejercicio**: Crea una columna `price_usd = price / rate` en `merged`.
- Implementa con `apply()` y luego con operación vectorizada. Compara tiempos.

```python
import time
start = time.time()
merged['p_usd_apply'] = merged.apply(lambda row: row['price']/row['rate'], axis=1)
t1 = time.time() - start
start = time.time()
merged['p_usd_vec'] = merged['price'] / merged['rate']
t2 = time.time() - start
t1, t2
```
