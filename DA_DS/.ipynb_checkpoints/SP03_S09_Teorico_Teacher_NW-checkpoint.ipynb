## 🐼 Introducción a Pandas

**Pandas** es una biblioteca de Python especializada en el análisis y manipulación de datos. Su nombre proviene de *"Panel Data"*, un término utilizado en econometría. Es una herramienta fundamental en ciencia de datos, análisis estadístico y procesamiento de información estructurada.

Con Pandas podemos:

- Cargar datos desde archivos como `.csv`, `.xlsx`, bases de datos, APIs, entre otros.
- Limpiar, filtrar y transformar datos de forma sencilla.
- Realizar análisis exploratorio mediante estadísticas descriptivas.
- Manipular series temporales y datos categóricos.
- Combinar, agrupar y reorganizar conjuntos de datos fácilmente.

Las dos estructuras de datos principales que ofrece Pandas son:

- **Series**: una lista unidimensional con etiquetas (índices).
- **DataFrames**: una tabla bidimensional con filas y columnas, similar a una hoja de cálculo o una tabla SQL.

A lo largo de esta clase aprenderemos a usar Pandas para explorar, limpiar y analizar datos reales de forma práctica.

### 3.1 Uso de `.loc[]` en Pandas

El método `.loc[]` permite **seleccionar y asignar** datos en un DataFrame utilizando **etiquetas** (índices y nombres de columna), en lugar de posiciones numéricas.

#### Sintaxis general
```python
df.loc[filtro_de_filas, selección_de_columnas]
```
Parámetros
- filtro_de_filas
    - Etiqueta única: una sola etiqueta de índice (df.loc[5]).
    - Lista de etiquetas: varias etiquetas (df.loc[[1, 3, 5]]).
    - Condición booleana: mascarilla que devuelve True/False por fila (df.loc[df["edad"] > 30]).
selección_de_columnas (opcional)
    - Etiqueta única: un nombre de columna (df.loc[:, "nombre"]).
    - Lista de etiquetas: varias columnas (df.loc[:, ["nombre", "edad"]]).
    - Rango de etiquetas: mediante slicing (df.loc[:, "col1":"col4"]).
    - Si se omite, se seleccionan todas las columnas (df.loc[filtro_de_filas]).

Ejemplos de uso

```python
# 1. Filtrar por condición y ver todas las columnas:
df.loc[df["ventas"] > 1000]

# 2. Filtrar filas y ver solo ciertas columnas:
df.loc[df["categoria"] == "Electrónica", ["producto", "precio"]]

# 3. Selección por etiquetas de fila y columna:
df.loc["fila_10", "columna_A"]

# 4. Asignación de valores a un subconjunto:
df.loc[df["puntuacion"] < 50, "estado"] = "revisar"
```

**Tip:** .loc respeta los nombres de índices y columnas. Para selección basada en posiciones, utiliza .iloc[].

````python
# Seleccionar la fila con índice 5 (sexta fila)
df.loc[5]
````

````python
# Seleccionar las filas con índices 2, 5 y 10
df.loc[[2, 5, 10]]
````

````python
# Seleccionar solo la columna 'Title'
df.loc[0:5, 'Title']
````

````python
# Seleccionar las filas 0 a 4 y las columnas 'Title' y 'Album'
df.loc[0:10, ['Title', 'Album']]
````

````python
# Una forma distinta de escribirlo
index_res = df.loc[5, 'Title': 'Album']
print(index_res)
````

### 3.2 Uso de `.iloc[]` en Pandas

```python
df.iloc[fila_inicio:fila_fin, columna_inicio:columna_fin]
```
- Qué hace: Selecciona filas y/o columnas por posiciones enteras (integer-location), ignorando las etiquetas del índice.
- Sintaxis básica:
    - df.iloc[5] → quinta fila (0-based) en todas las columnas.
    - df.iloc[2:8] → filas de la 3 a la 8 (excluye la 8).
    - df.iloc[:, 1] → segunda columna en todas las filas.
    - df.iloc[3:6, 0:2] → filas 4-6 y columnas 1-2.

**Uso típico:** Cuando se necesita seleccionar por posiciones numéricas o hacer slicing posicional en lugar de por etiquetas.

````python
df.iloc[5]
````

### 3.3 Slicing por posición en Pandas

```python
df[9:20]
Qué hace: Selecciona las filas cuya posición (integer-location) va de la 9 (incluida) hasta la 20 (excluida).
```

Equivalente a:

```
df.iloc[9:20]
```
**Uso típico:** Extraer un bloque continuo de filas para inspección o procesamiento.

````python
# de la 9 a la 20, incluidas
index_res = df[9:20]
print(index_res)
````

### 3.4 Operaciones de filtrado en Pandas

```python
df[df['Album'] == '1989']
```
- Selecciona todas las filas donde la columna **Album** es igual a `'1989'`.

```python
df[(df['Spotify Streams (millions)'] >= 900) & (df['Spotify Streams (millions)'] <= 1000)]
```
- Selecciona filas donde **Spotify Streams (millions)** está entre 900 y 1000 (inclusive).  
- Utiliza el operador **&** para combinar condiciones booleanas y paréntesis para agrupar cada condición.

> Estas técnicas se conocen como **indexación booleana** o **filtrado condicional**, y devuelven un DataFrame con solo las filas que cumplen los criterios especificados.


````python
# Filtrar canciones que pertenecen al álbum '1989'
df[df['Album'] == '1989']
````

````python
# Filtrar canciones con reproducciones entre 500 y 1000 millones
df[(df['Spotify Streams (millions)'] >= 900) & (df['Spotify Streams (millions)'] <= 1000)]
````

### 3.7 Operaciones de agrupamiento con `groupby` en Pandas

- **Qué hace**: Agrupa filas del DataFrame según valores de una o varias columnas, devolviendo un objeto `GroupBy` sobre el que se pueden aplicar funciones de agregación, transformación o filtrado.

- **Sintaxis básica**:
```python
df.groupby('columna')
df.groupby(['col1', 'col2'])
```

- **Métodos comunes**:
  - `agg()` o `aggregate()`: aplica funciones agregadas.
  - `mean()`, `sum()`, `count()`, `min()`, `max()`.
  - `size()`: devuelve el tamaño de cada grupo.
  - `get_group(nombre_grupo)`: obtiene el subconjunto correspondiente a un grupo específico.

- **Ejemplos**:
```python
# 1. Suma de ventas por categoría
df.groupby('categoria')['ventas'].sum()

# 2. Estadísticas descriptivas por región
df.groupby('region').describe()

# 3. Número de registros por usuario
df.groupby('usuario').size()

# 4. Aplicar múltiples agregaciones a ventas
df.groupby('categoria')['ventas'].agg(['sum', 'mean', 'max'])
```

````python
df.groupby('Release Year')['Title'].count()
````
