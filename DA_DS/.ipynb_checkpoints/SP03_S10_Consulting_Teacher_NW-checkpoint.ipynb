## Objetivos de la clase
- Cargar y explorar datos.
- Identificar valores faltantes.
- Filtrar datos y agrupar información.

```
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Carga de datos

```
df = pd.read_csv('/mnt/data/high_popularity_spotify_data.csv')
df.head()
```

La función `head()` nos permite ver las primeras filas del DataFrame.

## 2. Información general y estadísticos

```
df.info()

df.describe()
```

`info()` muestra tipos de datos y valores no nulos.
`describe()` proporciona estadísticas descriptivas de las variables numéricas.

## 3. Valores faltantes

```
df.isnull().sum()
```

Identificamos la cantidad de valores faltantes en cada columna.

## 4. Operaciones de filtrado

```
# Filtrar canciones con popularidad mayor a 90
popular_songs = df[df['popularity'] > 90]
popular_songs.head()
```

```
# Filtrar canciones de un artista específico (Taylor Swift)
taylor = df[df['artist_name'] == 'Taylor Swift']
taylor[['track_name', 'popularity']].head()
```

```
# Uso de loc para filtrar y seleccionar columnas
dance_rock = df.loc[(df['genre'] == 'rock') & (df['danceability'] > 0.7), ['track_name', 'popularity', 'danceability']]
dance_rock.head()
```

```
# Uso de iloc para obtener primeras 10 filas y primeras 5 columnas
df.iloc[0:10, 0:5]
```

## 5. Operaciones de agrupamiento (groupby)

```
# Agrupar por artista y calcular la popularidad media
artist_pop = df.groupby('artist_name')['popularity'].mean().sort_values(ascending=False)
artist_pop.head()
```

```
# Contar canciones por álbum
album_count = df.groupby('album_name')['track_name'].count().sort_values(ascending=False)
album_count.head()
```

```
# Agrupar por género y calcular la media de danceability
genre_dance = df.groupby('genre')['danceability'].mean()
genre_dance
```
