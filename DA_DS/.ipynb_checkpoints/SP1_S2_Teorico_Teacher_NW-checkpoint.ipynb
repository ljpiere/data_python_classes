## 1. Sesión de Breakout Rooms en Zoom

**Objetivo:** Practicar variables, manejo de cadenas, conversión de tipos y gestión de excepciones.

Divídanse en dos grupos y trabajen en los siguientes ejercicios durante 10 minutos. Luego compartiremos resultados.

- **Room 1:** Variables y conversiones

  1. Definan `x = "50"` (string) y `y = 20` (int).
  2. Conviertan `x` a entero y sumenlo a `y`.<br>
  3. Intenten dividir el resultado entre 0 dentro de un bloque `try/except`.

```python
# Solución Room 1
x = "50"
y = 20
try:
    total = int(x) + y
    print("Total:", total)
    _ = total / 0
except ZeroDivisionError:
    print("Error: división por cero detectada.")
```

- **Room 2:** Manejo de cadenas

  1. Tomen `saludo = "  ¡Hola Mundo!  "` y eliminen espacios.<br>
  2. Reemplacen "Mundo" por "Python" y dividan la frase con `split`.<br>
  3. Encadenen `.upper()` y `.replace()`.

```python
# Solución Room 2
saludo = "  ¡Hola Mundo!  "
s = saludo.strip()
f = s.replace("Mundo", "Python")
partes = f.split(' ')
mod = f.upper().replace('PYTHON', 'Data')
print(s, partes, mod)
```

## 2. Creación e Indexación de Listas

### Ejercicio 2.1 Básico
Crea `colores = ["rojo", "verde", "azul", "amarillo", "morado"]` y realiza:
1. Imprime el primer, tercer y último elemento.
2. Imprime el subgrupo de índices 1 a 3 (inclusive).

```python
# Solución Ejercicio 2.1
colores = ["rojo", "verde", "azul", "amarillo", "morado"]
print(colores[0])  # primero
print(colores[2])  # tercero
print(colores[-1]) # último
print(colores[1:4]) # índices 1,2,3
```

### Ejercicio 2.2 Índices Negativos y Largo
Usa la lista anterior:
1. Obtén el elemento en posición -2.
2. Imprime la longitud de la lista con `len`.
3. Intenta acceder a un índice fuera de rango y maneja el error con `try/except`.

```python
# Solución Ejercicio 2.2
colores = ["rojo", "verde", "azul", "amarillo", "morado"]
print(colores[-2])  # -2
print(len(colores))
try:
    print(colores[10])
except IndexError:
    print("Error: índice fuera de rango.")
```

## 3. Formateo de Cadenas dentro de Listas

### Ejercicio 3.1
Tienes `nombres = [" Ana", "Luis ", "  Carlos  "]`. Realiza:
1. Limpia cada nombre con `strip()`.
2. Crea mensajes: "Hola, Ana!", etc.
3. Almacena los mensajes en `saludos` y muéstralos.

```python
# Solución Ejercicio 3.1
nombres = [" Ana", "Luis ", "  Carlos  "]

nombre_0 = nombres[0].strip()
nombre_1 = nombres[1].strip()
nombre_2 = nombres[2].strip()

lista_nueva = [nombre_0, nombre_1, nombre_2]

print(f"Hola, {lista_nueva[0]}!")
```

### Ejercicio 3.2 Encadenamiento
Dada la lista `frases = ["uno,dos", "tres,cuatro"]`:
1. Para cada elemento, reemplaza ',' por ' y ' y luego conviértelo a mayúsculas.
2. Guarda resultados en `modificadas`.

```python
# Solución Ejercicio 3.2
frases = ["uno,dos", "tres,cuatro"]
modificadas = []
m = frases[0].replace(',', ' y ').upper()
modificadas.append(m)

n = frases[1].replace(',', ' y ').upper()
modificadas.append(n)
    
print(modificadas)
```

*Nota: Aunque en el ejercicio anterior usamos un bucle para mostrar, reemplázalo por dos líneas manuales para no usar loops.*

```python
# Solución sin bucle
frases = ["uno,dos", "tres,cuatro"]
mod1 = frases[0].replace(',', ' y ').upper()
mod2 = frases[1].replace(',', ' y ').upper()
modificadas = [mod1, mod2]
print(modificadas)
```

## 4. Funciones Integradas: `len`, `append` y `count`

Además de `len` y `append`, veamos `count`.

### Ejercicio 4.1 - `count`
Con `colores` lista de 5 colores, agrega otro "rojo". Luego usa `count` para saber cuántas veces aparece "rojo".

```python
# Solución Ejercicio 4.1
colores = ["rojo", "verde", "azul", "amarillo", "morado"]
colores.append("rojo")
print(colores.count("rojo"))
```

### Ejercicio 4.2 - Mezcla de conceptos
Usa la lista `valores = ["1", "2", "3"]`:
1. Convierte cada elemento a entero (sin usar bucles explícitos, asigna manualmente tres variables).<br>
2. Guarda en `nums` con `append`.
3. Imprime la suma de `nums` y su longitud.

```python
# Solución Ejercicio 4.2
v1 = int("1")
v2 = int("2")
v3 = int("3")
nums = []
nums.append(v1)
nums.append(v2)
nums.append(v3)
print(sum(nums), len(nums))
```
