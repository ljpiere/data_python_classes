# Clase: Introducción a Pandas con Taylor Swift Songs (Versión Extensa)
Exploración profunda del dataset `taylor_swift_songs.csv`.

## 1. Importaciones y configuración
Cargamos librerías y configuramos pandas para mostrar más información.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
pd.set_option('display.max_rows', 20)
pd.set_option('display.max_columns', None)
df = pd.read_csv('taylor_swift_songs.csv')
df.head()
````

## 2. Información básica y nulos
- Dimensiones
- Tipos de datos
- Valores nulos

````python
# Ejercicio 2.1
print('Dimensiones:', df.shape)
print(df.dtypes)
print('Nulos por columna:')
print(df.isnull().sum())
````

## 3. Estadísticas descriptivas
- `describe()`
- Percentiles personalizados

````python
# Ejercicio 3.1
df.describe()
# Ejercicio 3.2
df['Spotify Streams (millions)'].quantile([0.1, 0.25, 0.5, 0.75, 0.9])
````

## 4. Valores únicos y conteo
- Géneros y álbumes únicos
- `value_counts()`

````python
# Ejercicio 4.1
print('Géneros únicos:', df['Genre'].unique())
print(df['Genre'].value_counts())
# Ejercicio 4.2
print(df['Album'].value_counts())
````

## 5. Filtrado y ordenamiento
- Canciones con streams > 100M
- Top 10 más populares

````python
# Ejercicio 5.1
df[df['Spotify Streams (millions)'] > 100].sort_values(
    'Spotify Streams (millions)', ascending=False
).head(10)
````

## 6. Columnas derivadas
- Duración en segundos
- Bandera de popularidad

````python
# Ejercicio 6.1
df['Duration_sec'] = df['Duration (mins)'] * 60
# Ejercicio 6.2
mean_streams = df['Spotify Streams (millions)'].mean()
df['Is_Popular'] = df['Spotify Streams (millions)'] > mean_streams
df.head()
````

## 7. Agrupaciones y agregaciones
- Media de streams por género
- Duración promedio por año

````python
# Ejercicio 7.1
df.groupby('Genre')['Spotify Streams (millions)'].mean()
# Ejercicio 7.2
df.groupby('Release Year')['Duration_sec'].mean()
````

## 8. Operaciones con cadenas
- Contar coautores en `Songwriter(s)`
- Extraer primer productor

````python
# Ejercicio 8.1
df['Num_Songwriters'] = df['Songwriter(s)'].str.split(',').apply(len)
df[['Title','Num_Songwriters']].sort_values(
    'Num_Songwriters', ascending=False
).head()
# Ejercicio 8.2
df['First_Producer'] = df['Producer(s)'].str.split(',').str[0]
df[['Title','First_Producer']].head()
````

## 9. Pivot tables
- Streams promedio por género y año

````python
# Ejercicio 9
pd.pivot_table(
    df,
    values='Spotify Streams (millions)',
    index='Genre',
    columns='Release Year',
    aggfunc='mean'
)
````

## 10. Limpieza de datos
- Llenar nulos en `Awards`
- Renombrar columnas

````python
# Ejercicio 10
df_clean = df.fillna({'Awards':'None'})
df_clean = df_clean.rename(
    columns=lambda x: x.strip().lower().replace(' ', '_').replace('(', '').replace(')','')
)
df_clean.columns
````
